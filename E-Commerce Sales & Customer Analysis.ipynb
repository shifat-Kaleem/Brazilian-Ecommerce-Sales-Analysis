```python
import pandas as pd

orders = pd.read_csv("olist_orders_dataset.csv")
items = pd.read_csv("olist_order_items_dataset.csv")
products = pd.read_csv("olist_products_dataset.csv")
customers = pd.read_csv("olist_customers_dataset.csv")
sellers = pd.read_csv("olist_sellers_dataset.csv")
payments = pd.read_csv("olist_order_payments_dataset.csv")
reviews = pd.read_csv("olist_order_reviews_dataset.csv")
category = pd.read_csv("product_category_name_translation.csv")
```

```python
products[products["product_category_name"].isna()].head()
```

```python
category.head()
```

```python
category.columns
```

```python
products = products.merge(
    category,
    on="product_category_name",
    how="left"
)
```

```python
products.columns
```

```python
products[[
    "product_category_name",
    "product_category_name_english"
]].head(10)
```

```python
items.describe()
```

```python
items.columns
```

```python
items["shipping_limit_date"] = pd.to_datetime(items["shipping_limit_date"])
```

```python
items.info()
```

```python
items["total_item_value"] = items["price"] + items["freight_value"]
```

```python
items[["price", "freight_value", "total_item_value"]].head()
```

```python
print("Total Product Sales:", items["price"].sum())
print("Total Freight:", items["freight_value"].sum())
print("Total Value:", items["total_item_value"].sum())
```

```python
print("Average Product Price:", items["price"].mean())
print("Average Freight:", items["freight_value"].mean())
```

```python
sales = items.merge(
    products[["product_id", "product_category_name_english"]],
    on="product_id",
    how="left"
)
```

```python
sales.head()
```

```python
sales.shape
```

```python
category_sales = (
    sales.groupby("product_category_name_english")["price"]
    .sum()
    .sort_values(ascending=False)
)

category_sales.head(10)
```

```python
category_quantity = (
    sales.groupby("product_category_name_english")
    .size()
    .sort_values(ascending=False)
)

category_quantity.head(10)
```

```python
category_avg_price = (
    sales.groupby("product_category_name_english")["price"]
    .mean()
    .sort_values(ascending=False)
)

category_avg_price.head(10)
```

```python
date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for col in date_columns:
    orders[col] = pd.to_datetime(orders[col])
```

```python
orders["order_purchase_timestamp"].dtype
```

```python
orders["purchase_year"] = orders["order_purchase_timestamp"].dt.year
orders["purchase_month"] = orders["order_purchase_timestamp"].dt.month
orders["purchase_month_name"] = orders["order_purchase_timestamp"].dt.month_name()

orders["delivery_days"] = (
    orders["order_delivered_customer_date"] -
    orders["order_purchase_timestamp"]
).dt.days
```

```python
orders[[
    "order_id",
    "purchase_year",
    "purchase_month",
    "purchase_month_name",
    "delivery_days"
]].head()
```

```python
sales = sales.merge(
    orders[
        [
            "order_id",
            "order_status",
            "order_purchase_timestamp",
            "purchase_year",
            "purchase_month",
            "purchase_month_name",
            "delivery_days"
        ]
    ],
    on="order_id",
    how="left"
)
```

```python
sales.shape
```

```python
yearly_sales = (
    sales.groupby("purchase_year")["price"]
    .sum()
    .sort_index()
)

yearly_sales
```

```python
monthly_sales = (
    sales.groupby(["purchase_year", "purchase_month"])["price"]
    .sum()
    .sort_index()
)

monthly_sales
```

```python
yearly_orders = (
    sales.groupby("purchase_year")["order_id"]
    .nunique()
)

yearly_orders
```

```python
yearly_avg_order_value = yearly_sales / yearly_orders

yearly_avg_order_value
```

```python
delivery_analysis = sales[
    sales["order_status"] == "delivered"
]["delivery_days"]

delivery_analysis.describe()
```

```python
print("Average delivery days:", delivery_analysis.mean())
print("Median delivery days:", delivery_analysis.median())
print("Fastest delivery:", delivery_analysis.min())
print("Longest delivery:", delivery_analysis.max())
```

```python
sales = sales.merge(
    orders[
        [
            "order_id",
            "order_delivered_customer_date",
            "order_estimated_delivery_date"
        ]
    ],
    on="order_id",
    how="left"
)
```

```python
sales[
    [
        "order_delivered_customer_date",
        "order_estimated_delivery_date"
    ]
].head()
```

```python
delivered_orders = sales[sales["order_status"] == "delivered"].copy()

delivered_orders["delivery_delay_days"] = (
    delivered_orders["order_delivered_customer_date"]
    - delivered_orders["order_estimated_delivery_date"]
).dt.days
```

```python
delivered_orders["delivery_delay_days"].describe()
```

```python
print("Average delay:", delivered_orders["delivery_delay_days"].mean())
print("Median delay:", delivered_orders["delivery_delay_days"].median())
```

```python
sales = sales.merge(
    orders[["order_id", "customer_id"]],
    on="order_id",
    how="left"
)
```

```python
sales = sales.merge(
    customers[
        ["customer_id", "customer_unique_id", "customer_state"]
    ],
    on="customer_id",
    how="left"
)
```

```python
sales[[
    "customer_id",
    "customer_unique_id",
    "customer_state",
    "price"
]].head()
```

```python
state_sales = (
    sales.groupby("customer_state")["price"]
    .sum()
    .sort_values(ascending=False)
)

state_sales.head(10)
```

```python
state_customers = (
    sales.groupby("customer_state")["customer_unique_id"]
    .nunique()
    .sort_values(ascending=False)
)

state_customers.head(10)
```

```python
top_customers = (
    sales.groupby("customer_unique_id")["price"]
    .sum()
    .sort_values(ascending=False)
)

top_customers.head(10)
```
